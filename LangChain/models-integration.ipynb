```python
model = init_chat_model(
    "google_genai:gemini-3.5-flash-lite"
)
for chunk in model.stream("write a paragraph of 200 words about the history of Pakistan"):
    print(f"{chunk.text}\n")
```

```python
#BATCH (more then 1 prompt)
prompts = [
    "write a paragraph of 200 words about the history of Pakistan",
    "write a paragraph of 200 words about the history of India",
    "write a paragraph of 200 words about the history of Bangladesh",
]
responses = model.batch(prompts)
for response in responses:
    print(f"{response.text}\n")   
```
